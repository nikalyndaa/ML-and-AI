```python
"""
Завдання 4. Оцінювання стійкості моделі через крос-валідацію.

Використовуємо StandardScaler і LinearRegression у Pipeline,
а cross_val_score оцінює модель за R² на 5 частинах даних.
"""
```

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = make_regression(
    n_samples=2000,
    n_features=4,
    noise=15,
    random_state=42
)
model = make_pipeline(
    StandardScaler(),
    LinearRegression()
)
```

```python
scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring='r2'
)

mean_score = np.mean(scores)
std_score = np.std(scores)

print("Значення R² для кожного фолду:")
for fold_number, score in enumerate(scores, start=1):
    print(f"Фолд {fold_number}: {score:.6f}")

print(f"\nСереднє R²: {mean_score:.6f}")
print(f"Стандартне відхилення R²: {std_score:.6f}")
```
