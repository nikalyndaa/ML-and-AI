```python
from IPython.display import display, Markdown

temperature = [2,5,4,6,7,3,1]
avg = sum(temperature)/len(temperature)
maximum = max(temperature)
minimum = min(temperature)

dif = maximum - minimum

table = f"""| Показник | Значення |
|---|---:|
| Середнє значення | {avg:.2f} |
| Максимальна температура | {maximum} |
| Мінімальна температура | {minimum} |
| Різниця між максимальною та мінімальною | {dif} |
"""

display(Markdown(table))
          
```

```python
expenses  = [1200, 1500, 800, 600, 1000, 400, 900]
total = sum(expenses)
avg = total/len(expenses)
maximum = max(expenses)
minimum = min(expenses)
days_over_1000 = sum( 1 for expense in expenses if expense >1000)

table = f"""| Показник | Значення |
|---|---:|
| Загальна сума витрат | {total} |
| Середнє значення витрат | {avg:.2f} |
| Максимальна витрата | {maximum} |
| Мінімальна витрата | {minimum} |
| Кількість днів із витратами понад 1000 | {days_over_1000} |
"""

display(Markdown(table))
```

```python
from collections import Counter
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
counts = Counter(votes)
leader,max_count = counts.most_common(1)[0]

rows = "\\n".join(
    f"| {fruit} | {count} |"
    for fruit, count in sorted(counts.items(), key=lambda item: (-item[1], item[0]))
)
table = f"""| Фрукт | Кількість |
|---|---:|
{rows}
"""
display(Markdown(table))

print(f"Лідер: {leader} ({max_count} голосів)")

print("\nТоп-3:")
for place, (fruit, count) in enumerate(counts.most_common(3), start=1):
    print(f"{place}. {fruit} — {count}")
```

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Математика": math,
    "Фізика":physics,
    "Англійська":english}

rows=[]
for subject, scores in subjects.items():
    avg = sum(scores)/len(scores)
    minimum = min(scores)
    maximum = max(scores)
    rows.append(f"| {subject} | {avg:.2f} | {minimum} | {maximum} |")

table = """| Предмет | Середній бал | Мінімальний бал | Максимальний бал |
|---|---:|---:|---:|
""" + "\n".join(rows)

display(Markdown(table))

best_sub= max(subjects, key=lambda subject: sum(subjects[subject])/len(subjects[subject]))
print("Найкращий предмет: ",best_sub )
```
