```python
# Завдання 4. Додатковий аналіз біграм

import re
from collections import Counter

import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.util import bigrams

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

df = pd.read_csv("feedback.csv")
stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

all_bigrams = []

for review in df["feedback"].dropna():
    tokens = word_tokenize(clean_text(review))

    # Використовуємо ті самі правила очищення, що й у завданні 2
    filtered_tokens = [
        token for token in tokens
        if token not in stop_words and len(token) >= 4
    ]

    # Біграма — це послідовність із двох сусідніх слів
    all_bigrams.extend(bigrams(filtered_tokens))

bigram_counts = Counter(all_bigrams)
top_10 = bigram_counts.most_common(10)

print("10 найчастіших біграм:")
print(f"{'Біграма':<30} Частота")
print("-" * 42)

for pair, frequency in top_10:
    phrase = " ".join(pair)
    print(f"{phrase:<30} {frequency}")

# Формуємо таблицю та зберігаємо її у CSV
result = pd.DataFrame(
    [
        {"bigram": " ".join(pair), "frequency": frequency}
        for pair, frequency in top_10
    ]
)

result.to_csv("feedback_bigrams.csv", index=False, encoding="utf-8")
print("\nРезультати збережено у feedback_bigrams.csv.")
```
