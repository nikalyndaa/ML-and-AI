```python
# Завдання 2. Токенізація та видалення стоп-слів

import re
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Завантаження необхідних ресурсів NLTK (достатньо виконати один раз)
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

# Файл feedback.csv створюється скриптом task1_data_preparation.py
df = pd.read_csv("feedback.csv")

stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

all_tokens_before = []
all_tokens_after = []

for review in df["feedback"].dropna():
    cleaned_text = clean_text(review)

    # Розділяємо текст на окремі токени
    tokens = word_tokenize(cleaned_text)
    all_tokens_before.extend(tokens)

    # Видаляємо стоп-слова та токени довжиною менше 4 символів
    filtered_tokens = [
        token for token in tokens
        if token not in stop_words and len(token) >= 4
    ]
    all_tokens_after.extend(filtered_tokens)

print("Загальна кількість токенів до очищення:", len(all_tokens_before))
print("Загальна кількість токенів після очищення:", len(all_tokens_after))
print("\nПриклад токенів після очищення:")
print(all_tokens_after[:30])
```
