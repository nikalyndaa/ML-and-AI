```python
# Завдання 1. Підготовка та завантаження даних

import re
import pandas as pd

# 15 англомовних відгуків для аналізу
reviews = [
    "The product is excellent and works perfectly!",
    "Great quality, fast delivery, and friendly support.",
    "I really like this product. It is easy to use.",
    "The quality is good, but the price is a little high.",
    "Very happy with my purchase. I recommend it!",
    "The customer service was helpful and polite.",
    "Delivery was slow, but the product is excellent.",
    "This product is useful, reliable, and affordable.",
    "Amazing quality! I will definitely buy again.",
    "The product stopped working after two days.",
    "Good value for money and very easy to install.",
    "I expected better quality. The instructions were unclear.",
    "Excellent customer support and quick replacement.",
    "The design is beautiful and the product works well.",
    "Not bad, but the delivery could be faster."
]

# Створюємо CSV-файл із початковими відгуками
df = pd.DataFrame({"feedback": reviews})
df.to_csv("feedback.csv", index=False, encoding="utf-8")
print("Файл feedback.csv створено.")

# Завантажуємо дані за допомогою pandas
df = pd.read_csv("feedback.csv")

# Функція очищення: нижній регістр, видалення цифр і спецсимволів
def clean_text(text):
    text = text.lower()
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_feedback"] = df["feedback"].apply(clean_text)

# Кількість рядків і середня довжина відгуку у словах
df["word_count"] = df["clean_feedback"].apply(
    lambda text: len(text.split())
)

print("\nПерші очищені відгуки:")
print(df[["feedback", "clean_feedback"]].head())

print("\nКількість рядків:", len(df))
print("Середня довжина відгуку (слів):", round(df["word_count"].mean(), 2))

# Зберігаємо також очищені дані
df.to_csv("feedback_cleaned.csv", index=False, encoding="utf-8")
print("\nОчищені дані збережено у feedback_cleaned.csv.")
```
